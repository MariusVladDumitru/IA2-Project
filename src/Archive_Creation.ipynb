This notebook is not the main project notebook.  
The purpose of this notebook is to list the code used to create the dataset archives.  


For the archive **places2_random_flat_split.tar**:  
This archive contains a part of the places 2 dataset with images selected from both train and test splits.  
The original Places2 archive was downloaded from the following link: [https://www.kaggle.com/datasets/nickj26/places2-mit-dataset/data?select=train_256_places365standard](https://www.kaggle.com/datasets/nickj26/places2-mit-dataset/data?select=train_256_places365standard)  
I downloaded the original zip file(train_256_places365standard.zip) and used the script reduce_places2_zip_streaming.py to select images from it so that the total dimension of a new .tar (places2_random_flat) archive would be around 13GB. I originally intended for this new .tar archive to be uploaded to my google drive (that is why the almost 13GB dimension) but ended up uploading it to my student account one drive for download purposes in the main jupyter notebook.  
I used the following command to obtain **places2_random_flat.tar**:

```python
python reduce_places2_zip_streaming.py \
  --input-zip train_256_places365standard.zip \
  --output-tar places2_random_flat.tar \
  --target-gb 12.5 \
  --seed 42
```

Next i wanted to rename the remaining images into 0.<extension>, 1.<extension>, 2.<extension> so that, in the main notebook, indexing will be easier to do.  
Also, i split the images, into 2 subsets, train and val, 80% - train and 20% - val, to have the training and validation subsets.  
For this, i used the script split_tar_train_val_with_index.py. This also created the files train_index.json and val_index.json to make indexing easier, in the main notebook.  
I used the following command to obtain places2_random_flat_split.tar:

```python
python split_tar_train_val_with_index.py \
  --input-tar places2_random_flat.tar \
  --output-tar ./output/places2_random_flat_split.tar \
  --train-ratio 0.8 \
  --seed 42
```
